### User Class Explained
Definition: Represents a user that will be spawned with the responsibility to attack the system

Available Objects:
- abstract = True
- on_start()
- on_stop()
- tasks
- wait()
- wait_time()
- weight = 10

```python
from locust import User, task, constant


class MyScript(User):
    wait_time = constant(1)

    @task
    def launch(self):
        print("Launching the URL")

    @task
    def search(self):
        print("Searching")
```

### HTTP User Class
Definition: Represents an HTTP "user" that will be spawned with the responsibility to attack the system. Creating a client instance of HTTPSession
Methods Include:
- GET, POST, DELETE, PATCH, HEAD, PUT, headers, text, status_code, etc.
